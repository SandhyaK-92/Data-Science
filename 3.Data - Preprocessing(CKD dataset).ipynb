```python
from pandas.api.types import is_string_dtype
```

```python
def catNum(dataset):
    cat=[]
    num=[]
    

    for columnName in dataset.columns:
        
        if dataset[columnName].dtype == 'str' or str(dataset[columnName].dtype) == 'str':
            
           cat.append(columnName)
            
        else:
            
            num.append(columnName)
            
    return cat,num
```

```python
cat,num=catNum(dataset)
```

```python
cat
```

```python
num
```

```python
import numpy as np
from sklearn.impute import SimpleImputer
imp = SimpleImputer(missing_values=np.nan, strategy='most_frequent')
imp.fit(dataset[cat])
dataset[cat]=imp.transform(dataset[cat])
```

```python
import numpy as np
from sklearn.impute import SimpleImputer
imp = SimpleImputer(missing_values=np.nan, strategy='median')
imp.fit(dataset[num])
dataset[num]=imp.transform(dataset[num])
```
